3] Deep and shallow copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[0:5]
slice = slice * 10        # we are updating slice part   # and chnages are not affect main array so this is a deep copy
slice
```

```python
a   # after that we are checking like those changes are heppen in a array so those are not affected by slice part
    # but if there is a list then changes are happen
```

```python
# Deep copy :- separate copy. No change on original data.

# Shallow copy :- changes made will be reflected in original data too.
```

# **Advance Array Manipulation**

1] Stacking Array

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
a
```

```python
b
```

```python
np.vstack((a,b))  # vertical stack
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

2] Splitting Array

```python
c = np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,4) # horizontally spliting array in 4 parts # you need equal division at time of spliting
```

```python
np.vsplit(c,2)   # vertically spliting array in 2 parts  # you need equal division at time of spliting
```

```python
a = np.vsplit(c,4)
```

```python
for i in a:
  print(i)
```
